```python
# Simple BPE demonstration

words = ["low", "lower", "lowest", "low"]

# Initialize words as characters
vocab = [list(word) + ["</w>"] for word in words]

print("Initial character tokens:")
for word in vocab:
    print(word)

# Count adjacent pairs
from collections import Counter

pairs = Counter()

for word in vocab:
    for i in range(len(word) - 1):
        pairs[(word[i], word[i+1])] += 1

print("\nPair frequencies:")
for pair, count in pairs.items():
    print(pair, ":", count)

# Find most frequent pair
best_pair = pairs.most_common(1)[0]

print("\nMost frequent pair:", best_pair)
```

```python
from collections import Counter

words = ["low", "lower", "lowest", "low"]

# Initialize character tokens
vocab = [list(word) + ["</w>"] for word in words]

def get_pairs(words):
    pairs = Counter()

    for word in words:
        for i in range(len(word) - 1):
            pairs[(word[i], word[i+1])] += 1

    return pairs

def merge_pair(words, pair):
    merged_words = []

    for word in words:
        new_word = []
        i = 0

        while i < len(word):
            if i < len(word)-1 and (word[i], word[i+1]) == pair:
                new_word.append(word[i] + word[i+1])
                i += 2
            else:
                new_word.append(word[i])
                i += 1

        merged_words.append(new_word)

    return merged_words

# Perform 5 BPE merges
for step in range(5):
    pairs = get_pairs(vocab)

    if not pairs:
        break

    best_pair = pairs.most_common(1)[0][0]

    vocab = merge_pair(vocab, best_pair)

    print(f"Step {step+1}: Merge {best_pair}")
    print("Result:", vocab)
```

```python
from collections import Counter

class MyBPETokenizer:

    def __init__(self, num_merges=10):
        self.num_merges = num_merges
        self.merges = []
        self.vocab = {}

    def get_pairs(self, words):
        pairs = Counter()

        for word in words:
            for i in range(len(word) - 1):
                pairs[(word[i], word[i+1])] += 1

        return pairs

    def merge_pair(self, words, pair):
        result = []

        for word in words:
            new_word = []
            i = 0

            while i < len(word):
                if (i < len(word)-1
                    and (word[i], word[i+1]) == pair):

                    new_word.append(word[i] + word[i+1])
                    i += 2

                else:
                    new_word.append(word[i])
                    i += 1

            result.append(new_word)

        return result

    def train(self, text):
        words = text.split()
        vocab = [list(word) + ["</w>"] for word in words]

        for _ in range(self.num_merges):
            pairs = self.get_pairs(vocab)

            if not pairs:
                break

            best_pair = pairs.most_common(1)[0][0]

            vocab = self.merge_pair(vocab, best_pair)
            self.merges.append(best_pair)

        self.vocab = vocab

        print("Training completed!")
        print("Learned merges:", self.merges)


# Create tokenizer
tokenizer = MyBPETokenizer(num_merges=10)

# Train
tokenizer.train("low lower lowest low")
```

```python
class BPETokenizer:
    def __init__(self, merges):
        self.merges = merges

    def encode(self, text):
        tokens = list(text)

        while len(tokens) > 1:
            pairs = [
                (tokens[i], tokens[i+1])
                for i in range(len(tokens)-1)
            ]

            candidates = [
                (pair, i)
                for i, pair in enumerate(pairs)
                if pair in self.merges
            ]

            if not candidates:
                break

            best_pair = min(
                candidates,
                key=lambda x: self.merges[x[0]]
            )[0]

            new_tokens = []
            i = 0

            while i < len(tokens):
                if (i < len(tokens)-1 and
                    (tokens[i], tokens[i+1]) == best_pair):

                    new_tokens.append(
                        tokens[i] + tokens[i+1]
                    )
                    i += 2
                else:
                    new_tokens.append(tokens[i])
                    i += 1

            tokens = new_tokens

        return tokens

    def decode(self, tokens):
        return "".join(tokens)
```

```python
merges = {
    ('l', 'o'): 0,
    ('lo', 'w'): 1,
    ('low', 'e'): 2,
    ('lowe', 'r'): 3
}

tokenizer = BPETokenizer(merges)

text = "lower"

tokens = tokenizer.encode(text)
decoded = tokenizer.decode(tokens)

print("Original:", text)
print("Tokens:", tokens)
print("Decoded:", decoded)

assert decoded == text
```

```python
test_cases = [
    "hello world",
    "Large Language Models",
    "AI is amazing!",
    "12345",
    "Python, PyTorch."
]

for text in test_cases:
    tokens = tokenizer.encode(text)
    decoded = tokenizer.decode(tokens)

    print("\nOriginal:", text)
    print("Tokens:", tokens)
    print("Decoded:", decoded)

    assert decoded == text
```
